Q5. Why is dictionary lookup O(1) on average while searching for a value in a list using `in` is O(n)?Explain in plain English without using the word "hash" - then explain it again WITH the word "hash".


1.Without using the word “hash”

Dictionary lookup is O(1) on average because Python can go almost directly to the location where a key's data is stored. It does not need to check every item.

List search using in is O(n) because Python may have to check each element one by one until it finds the value.
example:
my_list = [10, 20, 30, 40, 50]
30 in my_list
2.With the word “hash”

A dictionary uses a hash table. Python applies a hash function to the key to calculate where the key-value pair should be stored. This allows Python to find the required item directly on average, giving O(1) lookup time.

A list does not use this mechanism. It searches elements one by one, so searching can take O(n) time.

Simple memory trick:

Dictionary → direct access → O(1) average
List → one-by-one search → O(n)

Why is NumPy significantly faster than Python lists for numerical operations? Give 3 reasons, atleast one should mention memory layout.  


NumPy is significantly faster than Python lists for numerical operations because:

Better memory layout: NumPy arrays store elements of the same data type in continuous (contiguous) memory, so the computer can access them efficiently.
Vectorized operations: NumPy can perform operations on many elements at once instead of using Python for loops for each element.
Optimized low-level code: Many NumPy operations are implemented in C/C/Fortran, which runs much faster than normal Python loops.

In short: NumPy is faster because it uses efficient memory storage, vectorization, and optimized low-level code.

```python
armstrong = []

for num in range(1, 10000):
    digits = str(num)
    n = len(digits)

    total = 0
    for digit in digits:
        total += int(digit) ** n

    if total == num:
        armstrong.append(num)

print("Armstrong numbers:", ", ".join(map(str, armstrong)))
print("Count:", len(armstrong))
```

```python
N = int(input("Enter N: "))

def is_prime(num):
    if num < 2:
        return False

    i = 2
    while i * i <= num:
        if num % i == 0:
            return False
        i += 1

    return True


if is_prime(N):
    print(N, "is Prime")
else:
    print(N, "is Not Prime")

prime_sum = 0

for num in range(2, N + 1):
    if is_prime(num):
        prime_sum += num

print("Sum of primes:", prime_sum)
```

```python
N = int(input("Enter a number: "))

if N < 0:
    print("Not Palindrome")
else:
    original = N
    reverse = 0

    while N > 0:
        digit = N % 10
        reverse = reverse * 10 + digit
        N = N // 10

    print("Reversed number:", reverse)

    if original == reverse:
        print("Palindrome")
    else:
        print("Not Palindrome")
```

```python
sentence = input("Enter a sentence: ")

characters = len(sentence)
words = 0
vowels = 0
consonants = 0
digits = 0
spaces = 0
special = 0

in_word = False

for ch in sentence:
    if ch.isalpha():
        if ch.lower() in "aeiou":
            vowels += 1
        else:
            consonants += 1

        if not in_word:
            words += 1
            in_word = True

    elif ch.isdigit():
        digits += 1
        in_word = False

    elif ch.isspace():
        spaces += 1
        in_word = False

    else:
        special += 1
        in_word = False

print("Total characters:", characters)
print("Total words:", words)
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special characters:", special)
```

```python
prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)

days = ("Monday", "Tuesday", "Wednesday",
        "Thursday", "Friday", "Saturday", "Sunday")

# (a) Maximum price and day
max_price = max(prices)
max_day = days[prices.index(max_price)]

# (b) Minimum price and day
min_price = min(prices)
min_day = days[prices.index(min_price)]

# (c) Average
average = sum(prices) / len(prices)

# (d) Biggest single-day gain
biggest_gain = prices[1] - prices[0]
gain_from = days[0]
gain_to = days[1]

for i in range(1, len(prices)):
    gain = prices[i] - prices[i - 1]

    if gain > biggest_gain:
        biggest_gain = gain
        gain_from = days[i - 1]
        gain_to = days[i]

# (e) Total volatility
volatility = 0

for i in range(1, len(prices)):
    volatility += abs(prices[i] - prices[i - 1])

print("Maximum price:", max_price, "on", max_day)
print("Minimum price:", min_price, "on", min_day)
print("Average price:", round(average, 2))
print("Biggest single-day gain:", biggest_gain,
      "between", gain_from, "and", gain_to)
print("Total volatility:", volatility)
```

```python
numbers = [
    10, 20, 30, 20, 40, 50, 10, 60, 70, 30,
    80, 90, 100, 50, 110, 120, 130, 140, 150, 160
]

# (a) Set of unique numbers
unique_set = set(numbers)

# Count each number without Counter
frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

# (b) Numbers appearing more than once
duplicates = []

# (c) Numbers appearing exactly once
single = []

for num in frequency:
    if frequency[num] > 1:
        duplicates.append(num)
    elif frequency[num] == 1:
        single.append(num)

print("Unique numbers:", unique_set)
print("Numbers appearing more than once:", duplicates)
print("Numbers appearing exactly once:", single)
```

```python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if gst_inclusive:
        total_amount = amount
        base_amount = amount / (1 + gst_rate / 100)
        gst_amount = total_amount - base_amount
    else:
        base_amount = amount
        gst_amount = amount * gst_rate / 100
        total_amount = base_amount + gst_amount

    return (
        round(base_amount, 2),
        round(gst_amount, 2),
        round(total_amount, 2)
    )


# Test cases
print(calculate_gst(1000))
print(calculate_gst(1180, gst_inclusive=True))
print(calculate_gst(500, gst_rate=5))
```

```python
def power(base, exp):
    if exp == 0:
        return 1

    if exp < 0:
        return 1 / power(base, -exp)

    return base * power(base, exp - 1)


# Test cases
print(power(2, 10))
print(power(5, 0))
print(power(3, -2))
```

```python
import numpy as np

# (a) Create a 5x5 identity matrix
matrix = np.eye(5)
print("Original Identity Matrix:")
print(matrix)

# (b) Multiply every element by 10
matrix = matrix * 10
print("\nAfter multiplying by 10:")
print(matrix)

# (c) Transpose the matrix
# The transpose of an identity matrix is itself because
# all non-zero elements are on the main diagonal.
transpose = matrix.T
print("\nTranspose of Matrix:")
print(transpose)

# (d) Sum of diagonal elements
diagonal_sum = np.trace(matrix)
print("\nSum of diagonal elements:", diagonal_sum)

# (e) Sum of all elements
total_sum = np.sum(matrix)
print("Sum of all elements:", total_sum)
```

```python
import numpy as np

sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])

days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])

# (a) Total weekly sales
total_sales = np.sum(sales)

# (b) Daily average
average_sales = np.mean(sales)

# (c) Day with maximum sales
max_day = days[np.argmax(sales)]
max_sales = np.max(sales)

# (d) Number of days above weekly average
days_above_average = np.sum(sales > average_sales)

# (e) Percentage growth from Monday to Sunday
growth = ((sales[-1] - sales[0]) / sales[0]) * 100

# (f) Day-on-day percentage changes
daily_changes = ((sales[1:] - sales[:-1]) / sales[:-1]) * 100

print("Total weekly sales: ₹", total_sales)
print("Daily average: ₹", round(average_sales, 2))
print("Maximum sales day:", max_day, "₹", max_sales)
print("Days above average:", days_above_average)
print("Monday to Sunday growth:", round(growth, 2), "%")
print("Day-on-day % changes:", np.round(daily_changes, 2))
```
